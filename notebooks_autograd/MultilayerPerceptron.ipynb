```python
import numpy as np
import matplotlib.pyplot as plt

from dlfs_autograd.layers import DenseLayer
from dlfs_autograd.activation import ReLU, Sigmoid
from dlfs_autograd.loss import MSE_Loss, BCE_Loss
from dlfs_autograd.optimizer import Optimizer_SGD
from dlfs_autograd.helpers import plot_2d_clf_problem
from dlfs_autograd.model import SequentialModel
```

# Simple XOR dataset for classification

```python
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

y = np.array([0, 1, 1, 0]).astype(np.int64)

plot_2d_clf_problem(X, y)
```

```python
layers = [
    DenseLayer(2, 8),
    Sigmoid(),
    DenseLayer(8, 1),
    Sigmoid()
]

model = SequentialModel(
    layers=layers,
    loss_function=BCE_Loss(), 
    optimizer=Optimizer_SGD(learning_rate=5e-1)
)

model.train(X, y.reshape(-1, 1), epochs=2000, print_every=500)
print(f'Accuracy: {np.mean(y.reshape(-1, 1) == np.round(model.predict(X)))}')
plot_2d_clf_problem(X, y, lambda x: model.predict(x) > 0.5)
```

```python
from sklearn.datasets import make_circles

X, y = make_circles(n_samples=200, factor=0.3, noise=0.1)
plot_2d_clf_problem(X, y)
```

```python
layers = [DenseLayer(2, 16), 
          ReLU(),
          DenseLayer(16, 16), 
          ReLU(), 
          DenseLayer(16, 1), 
          Sigmoid()]

model = SequentialModel(
    layers=layers, 
    loss_function=BCE_Loss(), 
    optimizer=Optimizer_SGD(learning_rate=5e-3, momentum=0.5, decay=1e-2)
)

model.train(X, y.reshape(-1, 1), print_every=100, epochs=500)
print(f'Accuracy: {np.mean(y.reshape(-1, 1) == np.round(model.predict(X)))}')
plot_2d_clf_problem(X, y, lambda x: model.predict(x) > 0.5)
```

```python
from sklearn.datasets import make_regression

X, y = make_regression(n_features=1, n_informative=1, n_targets=1, noise=5.3, random_state=52)

plt.scatter(X, y)
plt.show()
```

```python
layers = [DenseLayer(1, 1)]

model = SequentialModel(layers=layers, loss_function=MSE_Loss(), optimizer=Optimizer_SGD(learning_rate=1e-3, momentum=0.3, decay=0.))

model.train(X, y.reshape(-1, 1), print_every=500, epochs=1000)

plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red')
plt.show()
```
