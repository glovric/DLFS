```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

from dlfs_autograd.layers import MaxPoolLayer
from dlfs_autograd.tensor.base import Tensor
from .helpers import compare_arrays
```

```python
use_autograd = True

np.random.seed(0)
torch.manual_seed(0)

B = 2
C_in = 5
H_in = 18
W_in = 23

C_out = 9

kernel_size=2
stride=2
padding=0

x = Tensor(
    np.random.normal(size=(B, C_in, H_in, W_in))
)
print(x.shape)
```

```python
torch_x = torch.Tensor(x.data).requires_grad_(True)

torch_out = F.max_pool2d(torch_x, kernel_size, stride=stride)
print(torch_out.shape)
```

```python
dlfs_maxpool = MaxPoolLayer(pool_size=kernel_size, stride=stride, padding=padding)
dlfs_maxpool.forward(x)
dlfs_out = dlfs_maxpool.output
print(dlfs_out.shape)
```

```python
compare_arrays(torch_out.detach().numpy(), dlfs_out.data)
```

```python
grad = torch.ones_like(torch_out)
torch_out.backward(grad)
torch_grad = torch_x.grad.detach().numpy()
print(torch_grad.shape)
```

```python
dlfs_out.backward()
dlfs_grad = x.grad
print(dlfs_grad.shape)
```

```python
compare_arrays(torch_grad, dlfs_grad)
```
